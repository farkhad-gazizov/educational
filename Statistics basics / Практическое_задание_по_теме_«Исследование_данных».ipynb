Вычислим статистику для возраста клиентов банка

```python
import statistics as st
```

```python
# среднее арифметическое
avg_age = st.mean(df['Age'])
print(f'Среднее арифметическое возраста, лет: {round(avg_age, 2)}')
```

```python
# медиана
median_age = st.median(df['Age'])
print(f'Медиана возраста, лет: {round(median_age, 2)}')
```

```python
# стандартное отклонение
stdev_age = st.stdev(df['Age'])
print(f'Стандартное отклонение возраста, лет: {round(stdev_age, 2)}')
```

```python
# минимум
min_age = df['Age'].min()
print(f'Минимальный возраст, лет: {round(min_age, 2)}')
```

```python
# перцентили 25, 50, 75
df['Age'].quantile([0.25, 0.50, 0.75])
```

```python
# максимум
max_age = df['Age'].max()
print(f'Максимальный возраст, лет: {round(max_age, 2)}')
```

Визуализация плотности распределения возраста клиентов банка

```python
import matplotlib.pyplot as plt
```

```python
# Гистограмма плотности распределения возраста клиентов банка

plt.hist(df['Age'], 100)
plt.title('Распределение возраста клиентов банка')
plt.xlabel('Возраст')
plt.show()
```

```python
# График динамики значений
plt.plot(df['Age'])
plt.title('Динамика значений возраста')
plt.ylabel('Возраст')
plt.show()
```

```python
plt.boxplot(df['Age'])
plt.show()
```

**ВЫВОДЫ.**


1.   По гистограмме плотности распределения возраста клиентов банка делаем вывод, что имеем нормальное распределение. Это подверждается близкими значениями медианы и среднего арифметического.
2.   Основная доля распределения лежит в диапазоне среднее ± стандартное отклонение - 39.35 ± 10.81 лет, т.е. основная доля клиентов - люди среднего возраста.
3.   Самый молодой клиент - 7 лет, самый пожилой - 87 лет
4.   По графику динамики возраста и боксплоту видно, что имеется небольшое количество выбросов. Выбросы существенно не влияют на тип распределения и статистические показатели.
