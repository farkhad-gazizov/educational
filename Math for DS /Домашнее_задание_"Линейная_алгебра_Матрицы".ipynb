### Задание 1

 - Создайте матрицу 3 на 3, заполненную числами от 1 до 9
 - Прибавьте к ней матрицу с единицами на диагонале
 - Умножьте полученную матрицу на вектор (1, 2, 3)
 - Вычислить транспонированную матрицу $A^T$ от полученной в результате прибавления матрицы
 - Посчитайте обратную матрицу к полученной в результате прибавления матрице


````
import numpy as np
from numpy import linalg
````

````
m1 = np.arange(1, 10).reshape(3, 3)
m1
````

````
# прибавление единичной матрицы 3х3
m2 = m1 + np.eye(3)
m2
````

````
# умножение на вектор
m3 = m2 * np.array([1, 2, 3])
m3
````

````
# транспонирование
m4 = np.transpose(m2)
m4
````

````
# обратная матрица
m5 = np.linalg.inv(m2)
m5
````

````
# проверка, что m2 @ m5 даст единичную матрицу
m2 @ m5
````

````
# собственный вектор
# первый элемент в результате - собственные значения, остальные - собственные векторы (тут 2)
np.linalg.eig(m)
````

````
# приведем векторы к более красивому виду
res = np.linalg.eig(m)[1]
res
````

````
res2 = np.transpose(res)
res2
````

### Задание 3


Придумайте матрицу, у которой определитель = 1, и все элементы не равны 0.

P.S. Матрица должна быть ненулевой и размерностью не менее 2!

````
m = np.array([[1, 1], [1, 2]])
m
````

````
det = np.linalg.det(m)
det
````

````
import numpy as np

def find_matrix_with_det_1():
    for i in range(2, 100):
        for _ in range(100000):
            matrix = np.random.randint(1, 11, size=(i,i))
            det = np.linalg.det(matrix)
            # Сравнение с плавающей точкой через допуск 1e-8 = 0.00000001
            # Избегаем ошибок округления: det может быть 0.9999999999, а не 1.0
            if abs(det - 1.0) < 1e-8:
                print(matrix)
                return

find_matrix_with_det_1()
````
